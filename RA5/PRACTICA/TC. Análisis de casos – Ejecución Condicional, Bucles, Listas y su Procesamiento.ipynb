# Caso 1: Control de ventas diarias de una soda

## Problema
Una soda ubicada en San Pedro registra sus ventas de lunes a viernes. El programa debe calcular el total semanal, el promedio diario y clasificar el resultado según una meta.

## Entrada / datos
Use esta lista base:

```python
ventas_soda = [85000, 92000, 78000, 110000, 97000]
meta_semanal = 450000
```

## Proceso esperado

- Calcular total semanal.
- Calcular promedio diario.
- Comparar el total con la meta.
- Clasificar:
  - `Meta alcanzada` si el total es mayor o igual a la meta.
  - `Cerca de la meta` si el total alcanza al menos el 80% de la meta.
  - `Requiere atención` si no alcanza el 80%.

## Salida esperada
Debe mostrar total, promedio, porcentaje de cumplimiento y estado final.

## Pruebas manuales

| Escenario | Datos de entrada | Resultado esperado | Resultado obtenido |
|---|---|---|---|
| Caso normal | `[85000, 92000, 78000, 110000, 97000]`, meta `450000` | Meta alcanzada | |
| Caso límite | total exactamente igual a la meta | Meta alcanzada | |
| Caso especial | total menor al 80% de la meta | Requiere atención | |

````python
# Caso 1 - Escriba su solución aquí

# Complete el análisis
# Caso 1 - Escriba su solución aquí
# 1. Datos de entrada
ventas_soda = [85000, 92000, 78000, 110000, 97000]
meta_semanal = 450000

# 2. Procesamiento de datos
total_semanal = sum(ventas_soda)
promedio_diario = total_semanal / len(ventas_soda)
porcentaje_cumplimiento = (total_semanal / meta_semanal) * 100

# 3. Clasificación según la meta
if total_semanal >= meta_semanal:
    estado = "Meta alcanzada"
elif total_semanal >= (meta_semanal * 0.80):
    estado = "Cerca de la meta"
else:
    estado = "Requiere atención"

# 4. Salida de resultados
print("__REPORTE DE VENTAS SEMANAL____")
print(f"Total semanal: ₡{total_semanal:,.2f}")
print(f"Promedio diario: ₡{promedio_diario:,.2f}")
print(f"Porcentaje de cumplimiento: {porcentaje_cumplimiento:.2f}%")
print(f"Estado final: {estado}")
````

# Caso 2: Asistencia a una actividad cultural

## Problema
Una actividad cultural en Costa Rica registra asistencia por bloques horarios. El programa debe calcular la asistencia promedio y clasificar el nivel de ocupación.

## Entrada / datos
Use esta lista base:

```python
asistencias = [120, 180, 210, 250, 230]
capacidad = 300
```

## Proceso esperado

- Recorrer la lista con un ciclo `for`.
- Calcular asistencia total y promedio.
- Contar cuántos bloques superaron el 70% de capacidad.
- Clasificar el promedio de ocupación:
  - `Alta ocupación` si el promedio es mayor o igual al 80% de la capacidad.
  - `Ocupación media` si el promedio es mayor o igual al 50%.
  - `Baja ocupación` si el promedio es menor al 50%.

## Salida esperada
Debe mostrar total de asistentes, promedio, bloques sobre 70% y estado de ocupación.

## Pruebas manuales

| Escenario | Datos de entrada | Resultado esperado | Resultado obtenido |
|---|---|---|---|
| Caso normal | `[120, 180, 210, 250, 230]`, capacidad `300` | Ocupación media o alta según cálculo | |
| Caso límite | promedio exactamente 80% de capacidad | Alta ocupación | |
| Caso especial | lista con asistencias muy bajas | Baja ocupación | |

````python
# Caso 2 - Escriba su solución aquí

# Complete el análisis 
#  Datos_de_entrada
asistencias = [120, 180, 210, 250, 230]
capacidad = 300

# 2. Inicialización de variables para el ciclo
total_asistentes = 0
bloques_sobre_70 = 0
limite_70 = capacidad * 0.70

# Recorrido con ciclo for
for asistencia in asistencias:
    total_asistentes += asistencia
    if asistencia > limite_70:
        bloques_sobre_70 += 1

# 3. Cálculos de promedio y porcentaje
promedio_asistencia = total_asistentes / len(asistencias)
porcentaje_promedio = (promedio_asistencia / capacidad) * 100

# 4. Clasificación según nivel de ocupación
if promedio_asistencia >= (capacidad * 0.80):
    estado_ocupacion = "Alta ocupación"
elif promedio_asistencia >= (capacidad * 0.50):
    estado_ocupacion = "Ocupación media"
else:
    estado_ocupacion = "Baja ocupación"

# 5. Salida de resultados
print("=== REPORTE DE ASISTENCIA CULTURAL ===")
print(f"Total de asistentes: {total_asistentes}")
print(f"Promedio por bloque: {promedio_asistencia:.2f} ({porcentaje_promedio:.1f}% de capacidad)")
print(f"Bloques que superaron el 70% de capacidad: {bloques_sobre_70}")
print(f"Estado de ocupación: {estado_ocupacion}")
````

````python
# Caso 4 - Reto integrador

sedes = [
    {
        "nombre": "Soda San Pedro",
        "provincia": "San José",
        "tipo": "Alimentación",
        "ventas": [85000, 92000, 78000, 110000, 97000],
        "meta": 450000,
    },
    {
        "nombre": "Panadería Cartago",
        "provincia": "Cartago",
        "tipo": "Panadería",
        "ventas": [65000, 70000, 72000, 69000, 75000],
        "meta": 380000,
    },
    {
        "nombre": "Cafetería Liberia",
        "provincia": "Guanacaste",
        "tipo": "Cafetería",
        "ventas": [98000, 105000, 99000, 120000, 115000],
        "meta": 500000,
    },
]

## 1. FUNCIONES (Requisitos 1, 2 y 3)
# ==========================================

def calcular_total(ventas):
    """Calcula la suma total de las ventas."""
    return sum(ventas)

def calcular_promedio(ventas):
    """Calcula el promedio diario de ventas."""
    if not ventas:
        return 0
    return sum(ventas) / len(ventas)

def clasificar_sede(total, meta):
    """Clasifica el desempeño de la sede según su meta."""
    if total >= meta:
        return "Meta alcanzada"
    elif total >= (meta * 0.80):
        return "Cerca de la meta"
    else:
        return "Requiere atención"


# ==========================================
# 2. DATOS BASE
# ==========================================

sedes = [
    {
        "nombre": "Soda San Pedro",
        "provincia": "San José",
        "tipo": "Alimentación",
        "ventas": [85000, 92000, 78000, 110000, 97000],
        "meta": 450000,
    },
    {
        "nombre": "Panadería Cartago",
        "provincia": "Cartago",
        "tipo": "Panadería",
        "ventas": [65000, 70000, 72000, 69000, 75000],
        "meta": 380000,
    },
    {
        "nombre": "Cafetería Liberia",
        "provincia": "Guanacaste",
        "tipo": "Cafetería",
        "ventas": [98000, 105000, 99000, 120000, 115000],
        "meta": 500000,
    },
]


# 1. FUNCIONES (Requisitos 1, 2 y 3)


def calcular_total(ventas):
    """Calcula la suma total de las ventas."""
    return sum(ventas)

def calcular_promedio(ventas):
    """Calcula el promedio diario de ventas."""
    if not ventas:
        return 0
    return sum(ventas) / len(ventas)

def clasificar_sede(total, meta):
    """Clasifica el desempeño de la sede según su meta."""
    if total >= meta:
        return "Meta alcanzada"
    elif total >= (meta * 0.80):
        return "Cerca de la meta"
    else:
        return "Requiere atención"



# 2. DATOS BASE


sedes = [
    {
        "nombre": "Soda San Pedro",
        "provincia": "San José",
        "tipo": "Alimentación",
        "ventas": [85000, 92000, 78000, 110000, 97000],
        "meta": 450000,
    },
    {
        "nombre": "Panadería Cartago",
        "provincia": "Cartago",
        "tipo": "Panadería",
        "ventas": [65000, 70000, 72000, 69000, 75000],
        "meta": 380000,
    },
    {
        "nombre": "Cafetería Liberia",
        "provincia": "Guanacaste",
        "tipo": "Cafetería",
        "ventas": [98000, 105000, 99000, 120000, 115000],
        "meta": 500000,
    },
]



# 3. PROCESAMIENTO Y REPORTE (Requisitos 4, 5, 6 y 7)


provincias = set()
ranking = []
sede_lider = None
max_venta = -1


print("     REPORTE INTEGRAL DE DESEMPEÑO SEMANAL       ")


# Recorrido principal
for sede in sedes:
    nombre = sede["nombre"]
    provincia = sede["provincia"]
    total = calcular_total(sede["ventas"])
    promedio = calcular_promedio(sede["ventas"])
    cumplimiento = (total / sede["meta"]) * 100
    estado = clasificar_sede(total, sede["meta"])
    
    # Agregar provincia al set
    provincias.add(provincia)
    
    # Guardar en lista de tuplas para ranking
    ranking.append((nombre, total))
    
    # Identificar la sede con mayor venta total (Requisito 8)
    if total > max_venta:
        max_venta = total
        sede_lider = nombre

    # Imprimir reporte por sede
    print(f"📍 SEDE: {nombre.upper()}")
    print(f"   • Provincia: {provincia}")
    print(f"   • Tipo de negocio: {sede['tipo']}")
    print(f"   • Total semanal: ₡{total:,.2f}")
    print(f"   • Promedio diario: ₡{promedio:,.2f}")
    print(f"   • Cumplimiento de meta: {cumplimiento:.2f}%")
    print(f"   • Estado: [{estado}]")
    print("-" * 50)



# 4. RESUMEN GLOBAL Y RANKING

print("                 RESUMEN GLOBAL                   ")
print(f"Provincias con presencia ({len(provincias)}): {', '.join(provincias)}")
print(f"🏆 Sede líder en ventas: {sede_lider} (₡{max_venta:,.2f})")

print("\n--- Ranking Base (Tuplas) ---")
for pos, (nombre, total) in enumerate(ranking, start=1):
    print(f"{pos}. {nombre}: ₡{total:,.2f}")
````
